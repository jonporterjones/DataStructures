# Recursion examples

Fibonacci sequence

```python
def fib(n): 
   """
   Returns the n-th Fibonacci number.
   
   Order 2^n - exponential
   """
   print('fib')
   if n <= 1: 
       return n 
   else: 
       return fib(n - 1) + fib(n - 2)
```

```python
def fib(n, computed={0: 0, 1: 1}):
    """
    Returns the n-th Fibonacci number.

    Store any known Fibonacci numbers in computed.
    Order n

    Args:
        n (_type_): _description_
        computed (dict, optional): _description_. Defaults to {0: 0, 1: 1}.

    Returns:
        _type_: _description_
    """
    print('fib')
    if n not in computed:
        computed[n] = fib(n - 1, computed) + fib(n - 2, computed)
    return computed[n]
```

```python
fib(13)
```

Summing an array

```python
def arraySum(arr, index=0): 
    """
   We consider the sum of the array as the sum of the first element and the sum of the remaining elements.

    Args:
        arr (_type_): _description_
        index (int, optional): _description_. Defaults to 0.

    Returns:
        _type_: _description_
    """
    print(f'arraySum {arr} {index}')
    if index == len(arr): 
        return 0 
    else:
        return arr[index] + arraySum(arr, index + 1)
   
```

```python
arraySum([1,2,4])
```

Factorials

5! as 5 * 4! and 4! as 4 # 3! ... until 1! or 0!

```python
def factorial(n): 
   print(f"factorial {n}")
   if n == 0 or n == 1: 
       return 1
   else:
       return n * factorial(n - 1)
```

```python
factorial(5)
```

```python
def factorial(num):

    if num < 0:
        return "Error"
    elif num in {0, 1}:
        return num
    else:
        return num * factorial(num-1)

def factorials(nums):
    return [factorial(num) if factorial(num) is not None else 'Error' for num in nums]

print(factorials([2, 3, 4])) # should print: [2, 6, 24]
print(factorials([1, 5, 6])) # should print: [1, 120, 720]
print(factorials([0, -3, 10])) # should print: [1, 'Error', 3628800]
```

```python
# Some pretty slick results with factorial and memorization
def factorial(num, factorial_solved = {0: 0, 1: 1}):
    
    if num < 0:
        return "Error"
    else:
        if num not in factorial_solved:
            factorial_solved[num] = num * factorial(num - 1, factorial_solved)
            
    return factorial_solved[num]

def factorials(nums):
    results = []
    for num in nums:
        results.append(factorial(num))
        
    return results

print(factorials([2, 3, 4])) # should print: [2, 6, 24]
print(factorials([1, 5, 6])) # should print: [1, 120, 720]
print(factorials([0, -3, 10])) # should print: [1, 'Error', 3628800]
```
